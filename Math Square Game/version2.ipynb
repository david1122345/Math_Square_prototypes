```python
import random,numpy,time
import pygame
```

```python
class square():
    def __init__(self,x,y):
        self.height=0
        self.old_height=self.height
        self.clicked=((x>=25 and x<=75 and y>=25 and y<=75) and 0)
        self.x=x
        self.y=y
    def __repr__(self):
        return f"{self.height:.4}"
    def update(self):
        self.height-=1
        self.height=min(max(0,self.height),100)
        
    def set_oh(self):self.old_height=self.height
    @staticmethod
    def gc(h):
        r=2.5*h
        b=255-2.5*h
        return (int(r),0,int(b))
    def get_color(self):
       return square.gc(self.height)

class grid():
    def __init__(self,x_size,y_size):
        self.grid=[[square(i,j) for i in range(x_size)] for j in range(y_size)]
    def __repr__(self):
        s=""
        for j in self.grid:s+=j.__str__()+"\n"
        return s
    def get_blur_color(self,point_x,point_y):
        sum=0
        for t in range(-1,1):
            for l in range(-1,1):
                    try:
                        sum+=self.grid[point_x.__int__()+t][point_y.__int__()+l].height
                    except:pass
        return square.gc(sum/3**2)
    def update(self):
        for y in range(len(self.grid)):
            for x in range(len(self.grid[y])):
                self.grid[x][y].update()
                
                    

class bot():
    def __init__(self,goal):
        self.pos_x=random.randint(0,grid_size)
        self.pos_y=random.randint(0,grid_size)
        self.goal=goal
    def step(self,grid):
        if self.goal[0]==self.pos_x and self.goal[1]==self.pos_y:
            self.pos_x=random.randint(0,grid_size)
            self.pos_y=random.randint(0,grid_size)
        if self.goal[0]==self.pos_x or self.goal[1]==self.pos_y:
            if self.goal[0]==self.pos_x: self.pos_y+=int((self.goal[1]-self.pos_y)/abs((self.goal[1]-self.pos_y)))
            else:self.pos_x+=int(self.goal[0]-self.pos_x)/abs((self.goal[0]-self.pos_x))
            return
        sign_x=int(self.goal[0]-self.pos_x)/abs((self.goal[0]-self.pos_x))
        sign_y=int(self.goal[1]-self.pos_y)/abs((self.goal[1]-self.pos_y))
        sign_x=sign_x.__int__()
        sign_y=sign_y.__int__()
        self.pos_x=self.pos_x.__int__()
        self.pos_y=self.pos_y.__int__()
        try:
            if grid[self.pos_y][self.pos_x+sign_x].height>grid[self.pos_y+sign_y][self.pos_x].height: self.pos_x+=sign_x
            else:self.pos_y+=sign_y
        except Exception as a:
            self.pos_x=random.randint(0,grid_size)
            self.pos_y=random.randint(0,grid_size)
                    
grid_size=100
g7=700/grid_size
g=grid(grid_size,grid_size)
b=bot((23,23))
```

```python
pygame.init()
pygame.font.init()
font=pygame.font.SysFont('Arial',30)
WIDTH, HEIGHT = 700, 700
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Math Square")
running=True
fs=[]
while running:
        t=time.time()
        for event in pygame.event.get():
                if event.type==pygame.QUIT:
                        pygame.quit()
                        running=False
                        
                if event.type==pygame.MOUSEBUTTONDOWN:
                        pos_x,pos_y=event.pos
                        pos_x/=g7
                        pos_y/=g7
                        
                        for t in range(-10,10):
                                for l in range(-10,10):
                                        try:
                                                factor=100-numpy.sqrt((t**2+l**2))*4
                                                g.grid[pos_x.__int__()+t][pos_y.__int__()+l].height=int(factor)
                                        except:pass
                        b.goal=(int(pos_x),int(pos_y))
        for i in range(grid_size):
                for j in range(grid_size):
                        
                        c=g.grid[i][j].get_color()

                        pygame.draw.rect(screen,c,(g7*i,g7*j,g7,g7))
        pygame.draw.rect(screen,(245,30,0),(g7*b.pos_x,g7*b.pos_y,g7*3,g7*3))
        pygame.display.flip()
        g.update()
        b.step(g.grid)
        time.sleep(max(0.05-(time.time()-t),0))
```
