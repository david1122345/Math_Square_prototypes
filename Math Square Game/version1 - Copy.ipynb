```python
class square():
    def __init__(self,x,y):
        self.height=numpy.random.random()*100-50
        self.old_height=self.height
        self.clicked=((x>=25 and x<=75 and y>=25 and y<=75) and 0)
        self.x=x
        self.y=y
    def __repr__(self):
        return f"{self.height:.4}"
    def update(self,square):
        self.height+=(square.old_height-self.height-self.height)*0.8
        if abs(square.old_height+self.height-self.height)>45:self.height=square.old_height+self.height
        if ENTROPY:self.height-=numpy.random.rand()*10#entropy
        self.height=min(max(-50,self.height),50)
        
    def set_oh(self):self.old_height=self.height
    @staticmethod
    def gc(h):
        r=127+2.5*h
        b=200 if r<130 else 0
        #r-=r%15
        if r>100 and r<125:return (234,190,int(10+r*1.5))
        return (int(r)*0,int(r),int(b))
    def get_color(self):
       return square.gc(self.height)

class grid():
    def __init__(self,x_size,y_size):
        self.grid=[[square(i,j) for i in range(x_size)] for j in range(y_size)]
    def __repr__(self):
        s=""
        for j in self.grid:s+=j.__str__()+"\n"
        return s
    def get_blur_color(self,point_x,point_y):
        sum=0
        for t in range(-1,1):
            for l in range(-1,1):
                    try:
                        sum+=self.grid[point_x.__int__()+t][point_y.__int__()+l].height
                    except:pass
        return square.gc(sum/3**2)
    def update(self):
        for y in range(len(self.grid)):
            for x in range(len(self.grid[y])):
                self.grid[x][y].set_oh()
                
        for i in range(len(self.grid)):
            for j in range(len(self.grid)):
                    rl=list(range(8))
                    random.shuffle(rl)
                    for r in rl:
                        if j!=0 and r==0:self.grid[i][j].update(self.grid[i][j-1])
                        if i<len(self.grid)-1 and r==1:self.grid[i][j].update(self.grid[i+1][j])
                        if i!=0 and r==2:self.grid[i][j].update(self.grid[i-1][j])
                        if j<len(self.grid)-1 and r==3:self.grid[i][j].update(self.grid[i][j+1])
                        if j!=0 and i!=0 and r==4:self.grid[i][j].update(self.grid[i-1][j-1])
                        if j!=0 and i<len(self.grid)-1 and r==5:self.grid[i][j].update(self.grid[i+1][j-1])
                        if j<len(self.grid)-1 and i<len(self.grid)-1 and r==6:self.grid[i][j].update(self.grid[i+1][j+1])
                        if j<len(self.grid)-1 and i!=0 and r==7:self.grid[i][j].update(self.grid[i+-1][j+1])

class bot():
    def __init__(self,goal):
        self.pos_x=random.randint(0,grid_size)
        self.pos_y=random.randint(0,grid_size)
        self.goal=goal
    def step(self,grid):
        if self.goal[0]==self.pos_x and self.goal[1]==self.pos_y:
            self.pos_x=random.randint(0,grid_size)
            self.pos_y=random.randint(0,grid_size)
        if self.goal[0]==self.pos_x or self.goal[1]==self.pos_y:
            if self.goal[0]==self.pos_x: self.pos_y+=int((self.goal[1]-self.pos_y)/abs((self.goal[1]-self.pos_y)))
            else:self.pos_x+=int(self.goal[0]-self.pos_x)/abs((self.goal[0]-self.pos_x))
            return
        sign_x=int(self.goal[0]-self.pos_x)/abs((self.goal[0]-self.pos_x))
        sign_y=int(self.goal[1]-self.pos_y)/abs((self.goal[1]-self.pos_y))
        sign_x=sign_x.__int__()
        sign_y=sign_y.__int__()
        self.pos_x=self.pos_x.__int__()
        self.pos_y=self.pos_y.__int__()
        try:
            if grid[self.pos_y][self.pos_x+sign_x].height>grid[self.pos_y+sign_y][self.pos_x].height: self.pos_x+=sign_x
            else:self.pos_y+=sign_y
        except Exception as a:
            self.pos_x=random.randint(0,grid_size)
            self.pos_y=random.randint(0,grid_size)
                    
grid_size=100
g7=700/grid_size
ENTROPY=False
g=grid(grid_size,grid_size)
for  i in range(100):g.update()
b=bot((23,23))
ENTROPY=True
```
